```python
# TASK 3: Calculate the distances between a launch site to its proximities
from math import radians, sin, cos, sqrt, atan2

def haversine(coord1, coord2):
    # Coordinates in (latitude, longitude)
    R = 6371  # Earth radius in kilometers
    lat1, lon1 = coord1
    lat2, lon2 = coord2

    # Convert degrees to radians
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)

    a = sin(dlat / 2)**2 + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    return R * c  # Returns distance in kilometers
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property 
# for example
# distance_marker = folium.Marker(
#    coordinate,
#    icon=DivIcon(
#        icon_size=(20,20),
#        icon_anchor=(0,0),
#        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance),
#        )
#    )

import folium
from folium.features import DivIcon

# Coordinates
launch_site = (28.562302, -80.577356)
coastline_point = (28.56367, -80.57163)

# Calculate distance using haversine as fallback if geopy fails
from math import radians, sin, cos, sqrt, atan2

def haversine(coord1, coord2):
    R = 6371.0  # Earth radius in km
    lat1, lon1 = coord1
    lat2, lon2 = coord2

    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)

    a = sin(dlat/2)**2 + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c

distance = haversine(launch_site, coastline_point)

# Create site map if not yet created
site_map = folium.Map(location=launch_site, zoom_start=12)

# Add marker at coastline point
folium.Marker(
    location=coastline_point,
    popup='Closest Coastline',
    icon=folium.Icon(color='blue', icon='info-sign')
).add_to(site_map)

# Add distance marker between launch site and coastline
folium.Marker(
    location=[(launch_site[0] + coastline_point[0]) / 2,
              (launch_site[1] + coastline_point[1]) / 2],  # midpoint for clarity
    icon=DivIcon(
        icon_size=(150,36),
        icon_anchor=(0,0),
        html='<div style="font-size: 12px; color: #d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance),
    )
).add_to(site_map)

# Also mark the launch site
folium.Marker(
    location=launch_site,
    popup='Launch Site',
    icon=folium.Icon(color='green')
).add_to(site_map)

# Display map
site_map
```
